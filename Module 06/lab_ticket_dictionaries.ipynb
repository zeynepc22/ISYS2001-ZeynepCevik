## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns
  `0` for a category that was never recorded. It must not crash on a missing category.
- `add_transaction(spending, transaction)` takes one transaction, in the same shape as
  the records in the list above, and updates your dictionary so that amount is counted
  from then on.
- `summary(spending)` returns a dictionary that maps every category to its total.

How `spending` is built and shaped inside is entirely your decision. The checks only care
about what these three functions do, not how you store things underneath.

```python
# Build your dictionary from the transactions, and write your three functions here.

# total_for(spending, category)
# return amount spent or return 0 if category is non-existent.
def total_for(spending, category):
  return spending.get(category, 0)

# add_transaction(spending, transaction)
def add_transaction(spending, transaction):
  category = transaction['category']
  amount = transaction['amount']
  spending[category] = spending.get(category, 0) + amount

# summary(spending)
def summary(spending):
  return dict(spending)
```

```python
# Your own feature goes here.
# The feature I chose is to have a budget for types of spending and comparing the different types
spending = {}
for transaction in transactions:
  add_transaction(spending, transaction)


budget = {
    'rent': 1000,
    'groceries': 150,
    'transport': 50,
    'eating out': 50,
    'entertainment': 30,
    'phone bill': 30
}

def budget_Comparison(spending, budget):
  """Category -> (spend, limit, over / under). If categories are untracked, limit is None."""
  result = {}

  for category, spent in summary(spending).items():
    limit = budget.get(category)
    difference = None if limit is None else spent - limit
    result[category] = (spent, limit, difference)
  return result

# Loops through each category from the function before
# If no budget for that category, difference is None
# If budget is set, shows the money spent and its limit, whether you went over the budget or not.
for category, (spent, limit, difference) in budget_Comparison(spending, budget).items():
  if difference is None:
    # Add padding
    print(f"{category: <15} ${spent: >8} - No Budget Set :D")
  else:
    status = "Over" if difference > 0 else "Okay"
    print(f"{category: <15} ${spent: >8} / ${limit: <5} {status} by ${abs(difference)}")




```

## Self-check

Run this before you submit. Every line has to pass with no error, and the last line

```python
assert total_for(spending, 'groceries') == 165
assert total_for(spending, 'made up category') == 0

report = summary(spending)
assert report['rent'] == 1200
assert sum(report.values()) == 1682

before = total_for(spending, 'transport')
add_transaction(spending, {'date': '2026-08-30', 'category': 'transport',
                           'amount': 10, 'description': 'Bus top-up'})
assert total_for(spending, 'transport') == before + 10

print('All checks passed.')
```
